```python
!pip install pytorch-tabnet

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import RobustScaler, StandardScaler
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error, roc_auc_score
import xgboost as xgb
import lightgbm as lgb

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {device}")
```

```python
# Preprocessing for Deep Learning
feature_cols = ['cp_binary', 'gdp_per_capita', 'gini_index', 'rule_of_law', 'gender_inequality', 'education_index', 'log_gdp', 'cp_gdp_interaction', 'cp_gini_interaction', 'year']
target_homicide = 'log_homicide'
target_rape = 'log_rape'
target_cp = 'cp_binary'

# Prepare modeling dataframe
df_model = df_capital[feature_cols + [target_homicide, target_rape]].dropna()

X = df_model[feature_cols].values
y_hom = df_model[target_homicide].values
y_rape = df_model[target_rape].values
y_cp = df_model[target_cp].values

# Scaling
scaler_X = RobustScaler()
X_scaled = scaler_X.fit_transform(X)

# Split
X_train, X_test, y_hom_train, y_hom_test, y_rape_train, y_rape_test, y_cp_train, y_cp_test = train_test_split(
    X_scaled, y_hom, y_rape, y_cp, test_size=0.2, random_state=42
)

# To Tensors
X_train_t = torch.FloatTensor(X_train).to(device)
X_test_t = torch.FloatTensor(X_test).to(device)
y_hom_train_t = torch.FloatTensor(y_hom_train).reshape(-1, 1).to(device)
y_rape_train_t = torch.FloatTensor(y_rape_train).reshape(-1, 1).to(device)
y_cp_train_t = torch.FloatTensor(y_cp_train).reshape(-1, 1).to(device)

train_loader = DataLoader(TensorDataset(X_train_t, y_hom_train_t, y_rape_train_t, y_cp_train_t), batch_size=64, shuffle=True)
```

```python
class HybridCrimeModel(nn.Module):
    def __init__(self, input_dim):
        super().__init__()
        self.shared = nn.Sequential(
            nn.Linear(input_dim, 128), nn.ReLU(), nn.Dropout(0.2),
            nn.Linear(128, 64), nn.ReLU()
        )
        self.homicide_head = nn.Linear(64, 1)
        self.rape_head = nn.Linear(64, 1)
        self.cp_head = nn.Sequential(nn.Linear(64, 1), nn.Sigmoid())

    def forward(self, x):
        features = self.shared(x)
        return self.homicide_head(features), self.rape_head(features), self.cp_head(features)

model = HybridCrimeModel(len(feature_cols)).to(device)
optimizer = optim.AdamW(model.parameters(), lr=0.001)
criterion_reg = nn.MSELoss()
criterion_cls = nn.BCELoss()

# Training Loop (Simplified for execution)
for epoch in range(50):
    model.train()
    for b_x, b_hom, b_rape, b_cp in train_loader:
        optimizer.zero_grad()
        p_hom, p_rape, p_cp = model(b_x)
        loss = criterion_reg(p_hom, b_hom) + criterion_reg(p_rape, b_rape) + criterion_cls(p_cp, b_cp)
        loss.backward()
        optimizer.step()
    if epoch % 10 == 0: print(f"Epoch {epoch} complete.")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Ensemble Models: XGBoost & LightGBM
print("Training Ensemble Models...")
xgb_hom = xgb.XGBRegressor(n_estimators=100).fit(X_train, y_hom_train)
lgb_hom = lgb.LGBMRegressor(n_estimators=100, verbosity=-1).fit(X_train, y_hom_train)

# 2. Hybrid Model Predictions
model.eval()
with torch.no_grad():
    pred_hom_t, pred_rape_t, pred_cp_t = model(X_test_t)
    pred_hom = pred_hom_t.cpu().numpy()
    pred_rape = pred_rape_t.cpu().numpy()

# 3. Evaluation Metrics
print(f"\n--- HYBRID DL RESULTS ---")
print(f"Homicide R2: {r2_score(y_hom_test, pred_hom):.4f}")
print(f"Rape R2: {r2_score(y_rape_test, pred_rape):.4f}")

# 4. Visualization: Predicted vs Actual
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.scatter(y_hom_test, pred_hom, alpha=0.5, color='blue')
plt.plot([y_hom_test.min(), y_hom_test.max()], [y_hom_test.min(), y_hom_test.max()], 'r--')
plt.title('Homicide: Predicted vs Actual')

plt.subplot(1, 2, 2)
plt.scatter(y_rape_test, pred_rape, alpha=0.5, color='orange')
plt.plot([y_rape_test.min(), y_rape_test.max()], [y_rape_test.min(), y_rape_test.max()], 'r--')
plt.title('Rape: Predicted vs Actual')
plt.tight_layout()
plt.show()
```

```python
# 5. Counterfactual Analysis
print("Running Counterfactual Analysis...")
# Identify retentionist cases in the test set
retention_mask = (X_test[:, 0] > 0.5)
if retention_mask.any():
    X_cf = X_test[retention_mask].copy()
    X_cf[:, 0] = 0 # Simulate Abolition
    # Reset interaction terms related to CP
    X_cf[:, 7] = 0 # cp_gdp_interaction
    X_cf[:, 8] = 0 # cp_gini_interaction

    X_cf_t = torch.FloatTensor(X_cf).to(device)

    with torch.no_grad():
        orig_hom, _, _ = model(torch.FloatTensor(X_test[retention_mask]).to(device))
        cf_hom, _, _ = model(X_cf_t)

    avg_reduction = (cf_hom.mean() - orig_hom.mean()).item()
    print(f"Simulated Average Change in Log Homicide if abolished: {avg_reduction:.4f}")
else:
    print("No retentionist cases found in test split for simulation.")
```

Most standard international crime databases (like UNODC or World Bank) do not explicitly link a specific homicide to a preceding rape in their public CSVs. To answer this with 100% precision, we would typically need **Micro-data** (individual police reports) rather than **Macro-data** (national totals).

However, by looking at the ratio and the high-overlap countries above, we can identify regions where the risk of 'extreme' violent outcomes (where one crime leads to another) is statistically higher.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Using existing dataframes group_homicide_stats and group_rape_stats
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))

# Homicide plot
sns.barplot(data=group_homicide_stats, x='Group', y='homicide_total', palette='Reds_r', hue='Group', legend=False, ax=ax1)
ax1.set_title('Median Homicide Total (per 100k)', fontsize=13)
ax1.set_xlabel('')
ax1.set_ylabel('Homicides')

# Rape plot
sns.barplot(data=group_rape_stats, x='Group', y='rape_rate', palette='Blues_r', hue='Group', legend=False, ax=ax2)
ax2.set_title('Median Reported Rape Rate (per 100k)', fontsize=13)
ax2.set_xlabel('')
ax2.set_ylabel('Reported Rape Rate')

plt.suptitle('Comparison of Violent Crime Metrics by Country Grouping', fontsize=15, y=1.02)
plt.tight_layout()
plt.show()
```
